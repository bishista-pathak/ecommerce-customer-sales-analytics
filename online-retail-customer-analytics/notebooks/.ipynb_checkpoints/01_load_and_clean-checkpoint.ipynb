```python
# Importing Required Libraries

import pandas as pd
import numpy as np
```

```python
# Loading the Raw Online Retail Dataset

file_path = "../data/raw/online_retail_II.xlsx"

excel_file = pd.ExcelFile(file_path)
excel_file.sheet_names
```

## Loading the sheets

```python
# Loading the first sheet

df_2009_2010 = pd.read_excel(file_path, sheet_name="Year 2009-2010")
```

```python
# Loading the second sheet

df_2010_2011 = pd.read_excel(file_path, sheet_name="Year 2010-2011")
```

### Checking the sheet's dimensions

```python
print("--- 2009-2010 Dataset ---")
print("Shape:", df_2009_2010.shape)
```

```python
df_2009_2010.info()
```

```python
print("--- 2010-2011 Dataset ---")
print("Shape:", df_2010_2011.shape)
```

```python
df_2010_2011.info()
```

```python
# Checking whether both datasets have identical columns
columns_match = (df_2009_2010.columns.tolist() == df_2010_2011.columns.tolist())

print("Do both datasets have identical columns?", columns_match)
```

```python
# Combining the two datasets
df = pd.concat([df_2009_2010, df_2010_2011], ignore_index=True)
```

```python
print("Combined dataset shape:", df.shape)
```

### Initial inspection of the combined dataset

```python
print("--- First 5 Rows ---")
df.head()
```

```python
print("--- Dataset Information ---")
df.info()
```

```python
print("--- Descriptive Statistics ---")
df.describe(include="all")
```

```python
print("--- Missing Values ---")
df.isnull().sum()
```

```python
print("--- Duplicate Rows ---")
print(df.duplicated().sum())
```

### Inspect data quality issues

```python
# Check negative or zero quantities
print("--- Invalid Quantities ---")
print("Quantity <= 0:", (df["Quantity"] <= 0).sum())
```

```python
# Check negative or zero prices
print("--- Invalid Prices ---")
print("Price <= 0:", (df["Price"] <= 0).sum())
```

```python
# Check cancelled invoices
cancelled_invoices = (df["Invoice"].astype(str).str.startswith("C"))

print("Cancelled invoice rows:", cancelled_invoices.sum())
```

```python
# Check missing Customer IDs
print("--- Missing Customer IDs ---")
print("Missing CustomerID:", df["Customer ID"].isnull().sum())
```

### Clean the data

```python
# Rename column to remove the space (easier to work with)
df = df.rename(columns={'Customer ID': 'CustomerID'})
```

```python
# Remove rows with no CustomerID (we can't analyze customers we can't identify)
df = df.dropna(subset=['CustomerID'])
```

```python
# Remove cancelled invoices
df = df[~df["Invoice"].astype(str).str.startswith("C")]
```

```python
# Remove zero or negative quantities
df = df[df["Quantity"] > 0]
```

```python
# Remove zero or negative prices
df = df[df['Price'] > 0]
```

```python
# Remove exact duplicate rows
df = df.drop_duplicates()
```

### Feature Engineering

```python
# Calculate revenue for each transaction
df['Revenue'] = df['Quantity'] * df['Price']
```

```python
# Convert InvoiceDate to datetime
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])
```

```python
# Create Month column
df["Month"] = (df["InvoiceDate"].dt.to_period("M").astype(str))
```

```python
# Create DayOfWeek column
df["DayOfWeek"] = df["InvoiceDate"].dt.day_name()
```
