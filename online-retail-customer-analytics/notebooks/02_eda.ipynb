```python
# Import libraries

import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Load the cleaned dataset
df = pd.read_pickle("../data/processed/clean_data.pkl")
```

```python
# Check number of rows and columns
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
print("Dataset shape:", df.shape)
```

```python
# Display column names
print("Columns in the dataset:")
print(df.columns.tolist())
```

```python
# Display the first 5 rows
df.head()
```

```python
# Display the last 5 rows
df.tail()
```

```python
# Check data types and non-null counts
df.info()
```

```python
# Count missing values in each column
print("Missing values:")
df.isnull().sum()
```

```python
# Calculate missing-value percentage
missing_percentage = (df.isnull().mean() * 100).round(2)

print("Missing values (%):")
print(missing_percentage)
```

```python
# Check for duplicate rows
duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

```python
# Summary statistics for numerical columns
df.describe()
```

```python
# Check the time period covered by the dataset
start_date = df["InvoiceDate"].min()
end_date = df["InvoiceDate"].max()

print("Start date:", start_date)
print("End date:", end_date)
```

### Business Summary

```python
total_transactions = df["Invoice"].nunique()
total_customers = df["CustomerID"].nunique()
total_products = df["StockCode"].nunique()
total_countries = df["Country"].nunique()
total_quantity = df["Quantity"].sum()
total_revenue = df["Revenue"].sum()

print("Total transactions:", total_transactions)
print("Total customers:", total_customers)
print("Total products:", total_products)
print("Total countries:", total_countries)
print("Total quantity sold:", total_quantity)
print("Total revenue: £{:,.2f}".format(total_revenue))
```

### Calculate average transaction value

```python
# Average revenue per transaction
average_transaction_value = (total_revenue / total_transactions)

print("Average transaction value: £{:,.2f}".format(average_transaction_value))
```

### Analyze transaction volume by month

```python
# Number of transactions by month

monthly_transactions = (df.groupby("Month")["Invoice"].nunique())

print("Monthly transaction volume:")
print(monthly_transactions)
```

### Monthly revenue

```python
# Calculate total revenue for each month
monthly_revenue = (df.groupby("Month")["Revenue"].sum())

print("Monthly revenue:")
print(monthly_revenue)
```

### Plot monthly revenue

```python
# Plot monthly revenue trend

plt.figure(figsize=(10, 5))

plt.plot(monthly_revenue.index, monthly_revenue.values, marker="o")

plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue (£)")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

### Revenue by day of the week

```python
# Define the correct weekday order
day_order = [
    "Monday",
    "Tuesday",
    "Wednesday",
    "Thursday",
    "Friday",
    "Saturday",
    "Sunday"
]

# Calculate revenue by day
revenue_by_day = (df.groupby("DayOfWeek")["Revenue"].sum().reindex(day_order))

print("Revenue by day of the week:")
print(revenue_by_day)
```

### Visualize revenue by day

```python
plt.figure(figsize=(8, 5))

plt.bar(revenue_by_day.index, revenue_by_day.values)

plt.title("Revenue by Day of the Week")
plt.xlabel("Day of the Week")
plt.ylabel("Revenue (£)")
plt.xticks(rotation=30)

plt.tight_layout()
plt.show()
```

### Analyze countries

```python
print("Number of unique countries:", df["Country"].nunique())
```

```python
# Calculate revenue by country
country_revenue = (df.groupby("Country")["Revenue"].sum().sort_values(ascending=False))

print("Revenue by country:")
print(country_revenue)
```

```python
print(country_revenue.map(lambda x: f"£{x:,.2f}"))
```

### Find the UK's share of revenue

```python
uk_revenue = country_revenue["United Kingdom"]

uk_share = (uk_revenue / total_revenue)

print("UK revenue share: {:.1%}".format(uk_share))
```

### Top 10 countries excluding the UK

```python
top_countries = (country_revenue.drop("United Kingdom").head(10))

print("Top 10 countries by revenue excluding the UK:")
print(top_countries)
```

### Visualize top countries

```python
plt.figure(figsize=(9, 5))

plt.barh(top_countries.index[::-1], top_countries.values[::-1])

plt.title("Top 10 Countries by Revenue (Excluding UK)")
plt.xlabel("Revenue (£)")

plt.tight_layout()
plt.show()
```

### Product analysis

```python
print("Number of unique products:", df["StockCode"].nunique())
```

### Remove non-product entries

```python
# Exclude non-product transactions from product analysis
products_only = df[
    ~df["Description"].isin(["POSTAGE", "Manual"])
]

print("Rows after excluding non-product entries:", len(products_only))
```

### Top 10 products by revenue

```python
top_products_revenue = (
    products_only
    .groupby("Description")["Revenue"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

print("Top 10 products by revenue:")
print(top_products_revenue)
```

### Visualize top products

```python
plt.figure(figsize=(9, 5))

plt.barh(top_products_revenue.index[::-1], top_products_revenue.values[::-1])

plt.title("Top 10 Products by Revenue")
plt.xlabel("Revenue (£)")

plt.tight_layout()
plt.show()
```

### Top products by quantity sold

```python
top_products_quantity = (
    products_only
    .groupby("Description")["Quantity"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

print("Top 10 products by quantity sold:")
print(top_products_quantity)
```

### Visualize top products by quantity

```python
plt.figure(figsize=(9, 5))

plt.barh(
    top_products_quantity.index[::-1],
    top_products_quantity.values[::-1]
)

plt.title("Top 10 Products by Quantity Sold")
plt.xlabel("Quantity Sold")

plt.tight_layout()
plt.show()
```

### Final EDA summary

```python
eda_summary = pd.DataFrame({
    "Metric": [
        "Total Transactions",
        "Total Customers",
        "Total Products",
        "Total Countries",
        "Total Quantity Sold",
        "Total Revenue",
        "Average Transaction Value"
    ],
    
    "Value": [
        total_transactions,
        total_customers,
        total_products,
        total_countries,
        total_quantity,
        total_revenue,
        average_transaction_value
    ]
})

eda_summary["Value"] = eda_summary["Value"].map(lambda x: f"{x:,.2f}")

eda_summary
```
